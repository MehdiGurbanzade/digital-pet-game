````python
SCREEN_WIDTH = 1200
SCREEN_HEIGHT = 800
FPS = 60
STAT_DECAY_TIME = 120  # seconds
AGE_INCREASE_TIME = 300  # seconds


````

### Constants

In this part of the project, we define the **constants** that set the main rules for our digital pet game.  
Constants are values that do not change while the game runs. They help us control the size of the game window, the speed of the game, and how often the pet’s stats (like hunger, energy, and happiness) change. 

````python
import time

class Pet:
    def __init__(self):
        self.hunger = 100
        self.energy = 100
        self.happiness = 100
        self.health = 5
        self.age = 0
        self.last_decay = time.time()
        self.last_age_up = time.time()

    def decay_stats(self):
        now = time.time()
        if now - self.last_decay >= 120:
            self.hunger -= 5
            self.happiness -= 5
            self.last_decay = now
            self.check_health()

        if now - self.last_age_up >= 300:
            self.age += 1
            self.last_age_up = now

    def check_health(self):
        if self.hunger <= 0 or self.happiness <= 0 or self.energy <= 0:
            self.health -= 1
            self.hunger = max(self.hunger, 0)
            self.happiness = max(self.happiness, 0)
            self.energy = max(self.energy, 0)

    def feed(self):
        self.hunger = min(self.hunger + 5, 100)

    def play(self):
        self.happiness = min(self.happiness + 5, 100)
        self.hunger = max(self.hunger - 5, 0)
        self.energy = max(self.energy - 5, 0)

    def rest(self):
        self.energy = min(self.energy + 5, 100)
````

### 3. Create the Pet Class  

This part of the code defines the **Pet** class, which represents our digital pet *Alish*.  
It keeps track of important stats and provides actions the pet can take.  

#### Attributes:
- **hunger** → Starts at 100 (decreases over time or when playing).  
- **energy** → Starts at 100 (decreases during play, restored by resting).  
- **happiness** → Starts at 100 (decreases over time, increases when playing).  
- **health** → Starts at 5 hearts (decreases when hunger, happiness, or energy reach 0).  
- **age** → Starts at 0, increases every 5 minutes (300 seconds).  
- **last_decay / last_age_up** → Track the last time stats were updated or age increased.  

#### Methods:
- **decay_stats()** → Reduces hunger and happiness every 120 seconds. Also increases age every 300 seconds.  
- **check_health()** → Decreases health if hunger, energy, or happiness reach 0.  
- **feed()** → Increases hunger (max 100).  
- **play()** → Increases happiness but decreases hunger and energy.  
- **rest()** → Restores energy (max 100).  

This class acts as the **core engine** of the game, controlling how the pet lives, changes, and responds to the player’s actions.

````python
last_stat_decay_time = time.time()
last_age_up_time = time.time()
last_kitchen_time = None
last_bedroom_time = None
last_play_time = None
last_happiness_decay_time = time.time()


def draw_stats(pet):
    screen.blit(font.render(f"Age: {pet.age}", True, (255, 255, 0)), (10, 10))
    for i in range(pet.health):
        screen.blit(heart, (10 + i * 35, 50))

    def draw_bar(label, value, y_pos, color):
        max_width = 200
        bar_width = int((value / 100) * max_width)
        pygame.draw.rect(screen, (255, 255, 255), (10, y_pos, max_width, 25), 2)
        if value == 0:
            pygame.draw.rect(screen, (150, 0, 0), (10, y_pos, bar_width, 25))
        else:
            pygame.draw.rect(screen, color, (10, y_pos, bar_width, 25))
        text = font.render(f"{label}: {value}", True, (255, 255, 255))
        screen.blit(text, (220, y_pos - 5))

    draw_bar("Hunger", pet.hunger, 100, (255, 100, 100))
    draw_bar("Energy", pet.energy, 140, (100, 200, 255))
    draw_bar("Happiness", pet.happiness, 180, (255, 255, 100))
````

These variables make sure actions and changes happen at the correct intervals.

````python
def game_over_screen():
    
    screen.fill((0, 0, 0))
    title_font = pygame.font.SysFont(None, 64)
    button_font = pygame.font.SysFont(None, 48)
    msg = title_font.render("Game Over - Alish is gone", True, (255, 0, 0))
    restart_text = button_font.render("RESTART", True, (255, 255, 255))
    restart_rect = pygame.Rect(SCREEN_WIDTH // 2 - 100, SCREEN_HEIGHT // 2 + 20, 200, 60)
    pygame.draw.rect(screen, (0, 180, 0), restart_rect, border_radius=10)
    screen.blit(msg, (SCREEN_WIDTH // 2 - msg.get_width() // 2, SCREEN_HEIGHT // 2 - 60))
    screen.blit(restart_text, (restart_rect.centerx - restart_text.get_width() // 2, restart_rect.centery - restart_text.get_height() // 2))
    pygame.display.flip()
    waiting = True
    while waiting:
        for event in pygame.event.get():
            if event.type == pygame.QUIT:
                pygame.quit()
                sys.exit()
            if event.type == pygame.MOUSEBUTTONDOWN and restart_rect.collidepoint(event.pos):
                reset_game()
                waiting = False

````

### 6. Game Over Screen  

When Alish’s health reaches **0**, the game ends. To make this clear to the player, we show a **Game Over Screen**.  

#### Key Features:
1. **Black Background**
   - The entire screen is filled with black to signal the end of the game.  

2. **Game Over Message**
   - A large red text appears in the center:  
     `"Game Over - Alish is gone"`.  
   - This gives a clear message that the pet has passed away.  

3. **Restart Button**
   - A **green rounded rectangle button** is drawn with the label `"RESTART"`.  
   - The player can click this button to start over.  

4. **Event Handling**
   - The screen waits for user input:
     - If the player **closes the window**, the game exits.  
     - If the player **clicks the restart button**, the function `reset_game()` is called, and the game begins again.  

#### Why It’s Important:
This feature makes the game feel complete, giving players a chance to reflect on losing Alish but also encouraging them to **try again**. It prevents the game from just closing suddenly and adds a professional touch.  


````python
def reset_game():
    global pet, alish_rect, last_stat_decay_time, last_age_up_time, last_kitchen_time
    global last_bedroom_time, last_play_time, last_happiness_decay_time
    pet = Pet()
    alish_rect.center = (PLAY_AREA.centerx, SCREEN_HEIGHT - 100)
    last_stat_decay_time = time.time()
    last_age_up_time = time.time()
    last_kitchen_time = None
    last_bedroom_time = None
    last_play_time = None
    last_happiness_decay_time = time.time()

def show_start_screen():
    screen.fill((30, 30, 60))
    title_font = pygame.font.SysFont(None, 64)
    button_font = pygame.font.SysFont(None, 48)
    title = title_font.render("Welcome to Alish's World", True, (255, 255, 255))
    start_text = button_font.render("START", True, (255, 255, 255))
    start_rect = pygame.Rect(SCREEN_WIDTH // 2 - 100, SCREEN_HEIGHT // 2, 200, 60)
    pygame.draw.rect(screen, (0, 180, 0), start_rect, border_radius=10)
    screen.blit(title, (SCREEN_WIDTH // 2 - title.get_width() // 2, SCREEN_HEIGHT // 2 - 100))
    screen.blit(start_text, (start_rect.centerx - start_text.get_width() // 2, start_rect.centery - start_text.get_height() // 2))
    pygame.display.flip()
    waiting = True
    while waiting:
        for event in pygame.event.get():
            if event.type == pygame.QUIT:
                pygame.quit()
                sys.exit()
            if event.type == pygame.MOUSEBUTTONDOWN and start_rect.collidepoint(event.pos):
                waiting = False

show_start_screen()
````

### 7. Reset Game Function  

The `reset_game()` function is used whenever the player **loses and restarts** the game.  

#### What It Does:
1. **Recreates the Pet**
   - A new `Pet()` object is made so Alish starts fresh.  
   - Hunger, energy, happiness, health, and age are all reset to their starting values.  

2. **Resets Position**
   - Alish is placed back in the **Play Area**, standing at the bottom center.  

3. **Resets Timers**
   - All timers (`stat decay`, `age up`, `kitchen`, `bedroom`, `play area`, and `happiness decay`) are reset to the current time.  
   - This ensures stat changes start over correctly and fairly.  

✅ **Why It Matters**:  
Without this function, restarting the game would still use old data, which could cause bugs or unfair stat losses.

### 8. Start Screen  

The `show_start_screen()` function creates the **main menu** for the game.  

#### What It Does:
1. **Background**
   - Fills the screen with a dark blue color for a welcoming look.  

2. **Game Title**
   - Displays a large text:  
     `"Welcome to Alish's World"`  
   - This introduces the game to the player.  

3. **Start Button**
   - A green rounded rectangle labeled **START** is shown.  
   - Players can click it to begin the game.  

4. **Event Handling**
   - If the player **clicks START**, the waiting loop ends, and the game begins.

✅ **Why It Matters**:  
This gives the game a **professional introduction**, instead of throwing the player straight into gameplay. It sets the tone, makes the project feel polished, and helps beginners understand how to start.  


````python
while True:
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            pygame.quit()
            sys.exit()

    keys = pygame.key.get_pressed()
    if keys[pygame.K_LEFT]:                
        alish_rect.x -= 5
    if keys[pygame.K_RIGHT]:
        alish_rect.x += 5
    if keys[pygame.K_UP]:
        alish_rect.y -= 5
    if keys[pygame.K_DOWN]:
        alish_rect.y += 5

# 9. Player Movement  

# This section lets the player move "Alish" with the arrow keys.  

# What It Does:  
# 1. Left / Right Movement  
# - Pressing ← or → moves Alish 5 pixels left or right.  

# 2. Up / Down Movement  
# - Pressing ↑ or ↓ moves Alish 5 pixels up or down.  

    if alish_rect.left < 0: alish_rect.left = 0
    if alish_rect.right > SCREEN_WIDTH: alish_rect.right = SCREEN_WIDTH
    if alish_rect.top < SCREEN_HEIGHT // 2: alish_rect.top = SCREEN_HEIGHT // 2  
    if alish_rect.bottom > SCREEN_HEIGHT: alish_rect.bottom = SCREEN_HEIGHT
    current_time = time.time()
    decay_interval = 3 if PLAY_AREA.collidepoint(alish_rect.center) else 5

    if current_time - last_stat_decay_time >= decay_interval:
        pet.hunger = max(pet.hunger - 5, 0)
        pet.energy = max(pet.energy - 5, 0)
        pet.happiness = max(pet.happiness - 7, 0)
        pet.check_health()
        last_stat_decay_time = current_time

    if current_time - last_age_up_time >= 60:
        pet.age += 1
        last_age_up_time = current_time

    if pet.hunger < 40 or pet.energy < 40:
        if current_time - last_happiness_decay_time >= 1:
            pet.happiness = max(pet.happiness - 1, 0)
            last_happiness_decay_time = current_time
    elif current_time - last_happiness_decay_time >= 10:
        pet.happiness = max(pet.happiness - 5, 0)
        last_happiness_decay_time = current_time

    if KITCHEN.collidepoint(alish_rect.center):
        if last_kitchen_time is None or current_time - last_kitchen_time >= 1:
            pet.hunger = min(pet.hunger + 10, 100)
            
            last_kitchen_time = current_time
    else:
        last_kitchen_time = None

    if BEDROOM.collidepoint(alish_rect.center):
        if last_bedroom_time is None or current_time - last_bedroom_time >= 1:
            pet.energy = min(pet.energy + 10, 100)
            
            last_bedroom_time = current_time
    else:
        last_bedroom_time = None

    if PLAY_AREA.collidepoint(alish_rect.center):
        if last_play_time is None or current_time - last_play_time >= 1:
            pet.happiness = min(pet.happiness + 7, 100)
            pet.energy = max(pet.energy - 3, 0)
            pet.hunger = max(pet.hunger - 3, 0)
            last_play_time = current_time
    else:
        last_play_time = None

    if pet.hunger <= 0 or pet.energy <= 0 or pet.happiness <= 0:
        pet.health -= 1
        pet.hunger = 100
        pet.energy = 100
        pet.happiness = 100

    if pet.health <= 0:
        game_over_screen()


# 10. Health & Game Over  

# If core stats run out, Alish’s health decreases.  

# What It Does:  
# 1. Health Reduction  
# - If hunger, energy, or happiness hits 0, health decreases by 1.  
# - Stats reset back to 100 after losing health.  

# 2. Game Over Condition  
# - If health reaches 0, the Game Over screen is shown. 

    screen.blit(background, (0, 0))
    screen.blit(alish, alish_rect)

    if KITCHEN.collidepoint(alish_rect.center):
        screen.blit(apple, (alish_rect.centerx + 30, alish_rect.centery - 40))

    if PLAY_AREA.collidepoint(alish_rect.center):
        screen.blit(toy, (alish_rect.centerx + 30, alish_rect.centery - 40))

# 11. Drawing & Rendering  

# This section updates the visuals every frame.  

# What It Does:  
# 1. Background & Pet 
# - Draws the room background and Alish’s sprite.  

# 2. Item Display 
# - If Alish is in the kitchen, show an apple.  
# - If in the play area, show a toy.  
    

    draw_stats(pet)
    pygame.display.flip()
    clock.tick(60)
````
